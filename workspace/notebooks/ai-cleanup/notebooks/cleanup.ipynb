# AI Cleanup — standardized manifest

Columns must match the dashboard: `quantity`, `description`, `title`, …  

Put your downloaded CSV under `data/in/` (e.g. `data/in/order-316-cleanup.csv`).

```python
import sys
from pathlib import Path


def _bootstrap_notebooks_path() -> None:
    cwd = Path.cwd().resolve()
    for anchor in (cwd / "notebooks" / "_env.py", cwd.parent / "notebooks" / "_env.py"):
        if anchor.is_file():
            sys.path.insert(0, str(anchor.parent))
            return
    raise RuntimeError(
        "Could not find notebooks/_env.py — start Jupyter from workspace/notebooks/ai-cleanup (cwd="
        + str(cwd)
        + ")"
    )


_bootstrap_notebooks_path()
import _env  # noqa: E402
from ai_cleanup.io import load_manifest_csv, validate_manifest_columns

print("PROJECT_ROOT", _env.PROJECT_ROOT)
```

```python
manifest_path = _env.PROJECT_ROOT / "data" / "in" / "order-316-cleanup.csv"
if not manifest_path.is_file():
    raise FileNotFoundError(f"Add your standardized CSV here: {manifest_path}")

df = load_manifest_csv(manifest_path)
missing, extra = validate_manifest_columns(df)
display(df)
print("missing:", missing)
print("extra:", extra)
```
